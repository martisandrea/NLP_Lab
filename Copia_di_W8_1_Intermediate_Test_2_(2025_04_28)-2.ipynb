Assuming widow=5 (that is, +-2) and the following corpus:

new york is a place where every day feels like a new adventure. in new york, the streets are alive with new ideas, new faces, and new stories waiting to unfold. no matter how many times you visit new york, there’s always a new corner to explore and a new experience to chase. new york never grows old; it constantly reinvents itself.



- Compute the word-word matrix (i.e., term-context matrix) using raw counts. Only focus on actually computing the value for term=york and context=new
- Describe/show how to adapt the number above using tf-idf.


```python
import numpy  as np
v_1=np.array([35,84,49,0])
v_2=np.array([63,168,77,2])
v_3=np.array([42,0,210,70])

norm_v1=np.sqrt(v_1@v_1)
norm_v2=np.sqrt(v_2@v_2)
norm_v3=np.sqrt(v_3@v_3)
v1_v2= np.dot(v_1,v_2)/(norm_v1*norm_v2)
v1_v3= np.dot(v_1,v_3)/(norm_v1*norm_v3)
v2_v3= np.dot(v_2,v_3)/(norm_v2*norm_v3)
print(v1_v2)
print(v1_v3)
print(v2_v3)
#v3 è outliers
```
