```python


import numpy as np
#Q1 Logistic Regression w = [2, -1, 0.5], x = [1, 3, 2],
#b = 0 Calcolare: z, σ(z), classe

w=np.array([2, -1, 0.5])
x=np.array([1, 3, 2])
b=0
z=w@x+b
sigm=1/(1 + np.exp(-z))
print(sigm)
if sigm >= 0.5:
    print('Classe positiva')
else:
    print('Classe negativa')
```

```python
#Q5 Vector similarity v1=(1,2,3), v2=(2,4,6), v3=(0,1,10) Dot product, outlier

import numpy as np
v1= np.array([1 ,2 ,3 ])
v2= np.array([2,4,6])
v3= np.array([0,1,10])

v1_norm= np.sqrt(v1@v1)
v2_norm = np.sqrt(v2@v2)
v3_norm = np.sqrt(v3@v3)
v1v2= np.dot(v1,v2)/v1_norm/v2_norm
v1v3= np.dot(v1,v3)/v1_norm/v3_norm
v3v2= np.dot(v3,v2)/v3_norm/v2_norm
print(v1v2)
print(v1v3)
print(v3v2)
#v3
```

```python
#Q6 Word-context matrix Corpus: 'the cat was small the cat was fast' Window=±2 Conteggio (cat,
#was)
corpus="the cat was small the cat was fast"
count=0
window_size=2
tokens=nltk.regexp_tokenize(corpus, r"\w+")
for i,t in enumerate(tokens):
  if t=="cat":

    start=max(0, i-window_size)
    end=min(len(tokens), i+window_size+1)

    window = tokens[start:i] + tokens[i+1:end]
    for w in window:
      if w == "was":
        count+=1
print(count)
```

```python
#SGD x1=2, x2=1, y=1, w1=0, w2=0, b=0, η=0.1 Calcolare σ, gradienti, aggiornare parametri



y=1
x = np.array([2,1])
w = np.array([0,0])
b=0
lr=0.1
#forward pass
z=np.dot(x,w) + b
sigm=1/(1+np.exp(-z))
print(sigm)
#backprop

#derivata
delta = sigm - y
# Calcolo Gradienti
grad_w = delta * x
grad_b = delta
print(grad_w)
print(grad_b)
#finebackprop
#aggiornamento parametri
w = w - lr * grad_w
b = b - lr * grad_b
print(w)
print(b)
```
