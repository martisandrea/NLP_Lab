## Q1 probabilities with n-gram models (6 points)

Given the following knowledge (ngram, count):


    (advances in natural) 2
    (advances in) 38
    (in natural) 54
    (advances) 129
    (in) 25903
    (natural) 470
    (*) 100000   # corpus size

What is the probability that a sentence _"In recent years there has been several advances in "_ will continue with _"natural"_

Compute the probability using:
 - 3-grams
 - bigrams
 - unigrams
 - stupid backoff (assuming that you apply it recursively whenever count is `<10`)
 - interpolation with weights: `W_tri=0.6`, `W_bi=0.3`, `W_uni=0.1`

```python
# Trigrams
prob_trigrams = 2/38
print(f"Probability of trigrams: {prob_trigrams}")

# Bigrams
prob_bigrams = 54/25903
print(f"Probability of bigrams: {prob_bigrams}")

# Unigrams
prob_unigrams = 470/100000
print(f"Probability of unigrams: {prob_unigrams}")

# Stupid Backoff - solo uno step perchè Count(trigram) =2 < 10 ma Count(bigram) 54 >10
S_trigram = 0.4 * prob_bigrams
print(f"Probability of Stupid Backoff: {S_trigram}")

# Interpolation
I_trigram = 0.6 * prob_trigrams + 0.3 * prob_bigrams + 0.1 * prob_unigrams
print(f"Probability of Interpolation: {I_trigram}")
```

## Q3 practical analysis on a corpus (6 points)
Using the "carroll-alice" corpus of the Gutenberg project, in Python or using bash command line compute the following:
- |V|
- |V| with normalization based on removing capitalization
- get a random word and compute its count and probability (either with and without normalization)
- get a bigram and and compute its count and probability (either with and without normalization)

You can use the code below to access from python and/or save the corpus to a file.

```python
!pip install nltk

fname = 'carroll-alice.txt'

import nltk
nltk.download('gutenberg')
from nltk.corpus import gutenberg


corpus = gutenberg.raw(fname)                  # corpus can be used on python
with open(fname, "w", encoding="utf-8") as f:  # save to file for bash commands
    f.write(corpus)
```

```python
import random
vocabulary = nltk.regexp_tokenize(corpus, r"\w+")
vocabulary_norm = nltk.regexp_tokenize(corpus.lower(), r"\w+")

v_size = len(set(vocabulary))
v_size_norm = len(set(vocabulary_norm))

print(f"Vocabulary: {v_size}")
print(f"Vocabulary with normalization: {v_size_norm}")

random_word = random.choice(list(vocabulary_norm))
print("Random word: " + random_word)

count_random_word = vocabulary.count(random_word)
count_random_word_norm = vocabulary_norm.count(random_word)
print(f"Count of random word: {count_random_word}")
print(f"Count of random word (norm): {count_random_word_norm}")

prob_random_word = count_random_word/len(vocabulary)
prob_random_word_norm = count_random_word_norm/len(vocabulary_norm)
print(f"Probability of random word: {prob_random_word}")
print(f"Probability of random word (norm): {prob_random_word_norm}")

# bigrammi
id = random.randint(0, len(vocabulary))
bigram_random = vocabulary_norm[id] + " " + vocabulary_norm[id+1]
print("Bigram: " + bigram_random)
bigrams = []
for i in range(len(vocabulary)-1):
  bigrams.append(vocabulary[i] + " " + vocabulary[i+1])
bigrams_norm = []
for i in range(len(vocabulary_norm)-1):
  bigrams_norm.append(vocabulary_norm[i] + " " + vocabulary_norm[i+1])

count_bigram_random = bigrams.count(bigram_random)
count_bigram_random_norm = bigrams_norm.count(bigram_random)
prob_bigram_random = (count_bigram_random + 1)/(vocabulary.count(vocabulary_norm[id]) + v_size)
prob_bigram_random_norm = (count_bigram_random_norm + 1)/(vocabulary_norm.count(vocabulary_norm[id]) + v_size_norm)

print(f"Probability of bigram: {prob_bigram_random}")
print(f"Probability of bigram (norm): {prob_bigram_random_norm}")
```

##Q4 (6 points)

Given 3 semantic concepts associated to the follwing vectors:

    v_1 =(5,12,7,3)
    v_2 =(9,24,11,2)
    v_3 =(6,0,30,10)

 - Show which one is the outlier (the most different in semantics).
 - are they expected to be raw counts, tf-idf or word2vec embeddings?

```python
# cosine similarity
import numpy as np
v1 = np.array([5,12,7,3])
v2 = np.array([9,24,11,2])
v3 = np.array([6,0,30,10])

cos_sim_1_2 = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
cos_sim_1_3 = np.dot(v1, v3) / (np.linalg.norm(v1) * np.linalg.norm(v3))
cos_sim_2_3 = np.dot(v2, v3) / (np.linalg.norm(v2) * np.linalg.norm(v3))

print(f"Cosine similarity between v1 and v2: {cos_sim_1_2}")
print(f"Cosine similarity between v1 and v3: {cos_sim_1_3}")
print(f"Cosine similarity between v2 and v3: {cos_sim_2_3}")
```

##Q5 (6 points)
Assuming window=5 (that is, +-2) and the following corpus:

The city was quiet that morning. The city streets, usually crowded and noisy, felt empty and calm. As I was walking through the city, I noticed the city parks, even a small coffee shop was there. All of them seemed to be holding their breath, waiting for the city to wake up again

- Compute the word-word matrix (i.e., term-context matrix) using raw counts. Only focus on actually computing the value for term=city and context=was
- Show how to adapt the number above using tf-idf.


```python
import nltk
import numpy as np

corpus = "The city was quiet that morning. The city streets, usually crowded and noisy, felt empty and calm. As I was walking through the city, I noticed the city parks, even a small coffee shop was there. All of them seemed to be holding their breath, waiting for the city to wake up again"
tokens=nltk.regexp_tokenize(corpus,r"\w+")
count = 0
for i,word in enumerate(tokens):
  if word == "city":
    start = max(0, i-2)
    end = min(len(tokens), i+3)
    if "was" in tokens[start:end]:
      count+=1

print(f"Number of times that 'was' is contained in the window having 'city' as center: {count}")

#tf-idf
tf_idf_city = np.log10(count + 1) * np.log10(1)
print(f"TF-IDF normalization of counts: {tf_idf_city}")
```
