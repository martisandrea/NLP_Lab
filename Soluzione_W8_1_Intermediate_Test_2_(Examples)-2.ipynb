```python
import numpy as np

v1 = np.array([5, 12, 7, 3])
v2 = np.array([9, 24, 11, 2])
v3 = np.array([6, 0, 30, 10])


v1_norma = np.sqrt(v1 @ v1)
v2_norma = np.sqrt(v2 @ v2)
v3_norma = np.sqrt(v3 @ v3)

v1v2= np.dot(v1,v2)/v1_norma/v2_norma
v1v3= np.dot(v1,v3)/v1_norma/v3_norma
v2v3= np.dot(v2,v3)/v2_norma/v3_norma

print(v1v2)
print(v1v3)
print(v2v3)
```
